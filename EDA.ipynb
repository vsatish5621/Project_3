Outlier Detection

Outlier detection was performed to identify unusually high or low values in numerical variables. Outliers are important because they can significantly influence statistical measures and business interpretations.

In this project, outliers were detected using:

Boxplots
Interquartile Range (IQR) Method

The analysis focused on:

employee_count
annual_revenue_million
customer_count

Since the dataset represents real-world companies, extreme values may correspond to large multinational organizations. Therefore, outliers were analyzed but not removed.

```python
Q1 = df['annual_revenue_million'].quantile(0.25)
Q3 = df['annual_revenue_million'].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

revenue_outliers = df[
    (df['annual_revenue_million'] < lower_limit) |
    (df['annual_revenue_million'] > upper_limit)
]

print("Revenue Outliers:", len(revenue_outliers))
```

The Interquartile Range (IQR) method was used to identify revenue outliers.

The lower and upper boundaries were calculated using:

Lower Bound = Q1 − 1.5 × IQR
Upper Bound = Q3 + 1.5 × IQR

Companies falling outside these boundaries were considered outliers.

```python
# Employee Count Outliers
Q1 = df['employee_count'].quantile(0.25)
Q3 = df['employee_count'].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

employee_outliers = df[
    (df['employee_count'] < lower_limit) |
    (df['employee_count'] > upper_limit)
]

print("Employee Outliers:", len(employee_outliers))
```

```python
#Customer count Outliers
Q1 = df['customer_count'].quantile(0.25)
Q3 = df['customer_count'].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

customer_outliers = df[
    (df['customer_count'] < lower_limit) |
    (df['customer_count'] > upper_limit)
]

print("Customer Outliers:", len(customer_outliers))
```

Conclusion for Outlier Detection: 
Outlier detection identified a number of extreme observations in revenue, employee count, and customer count. These values were retained because they represent genuine large-scale enterprises and multinational organizations. Removing such observations could result in the loss of valuable business information.

# Bivariate Analysis

Introduction

Bivariate analysis examines the relationship between two variables. The objective is to identify patterns, trends, and associations that influence business performance.

Bivariate analysis examines the relationship between two variables to identify patterns, dependencies, and business trends.



```python
#Industry vs Average Revenue

industry_rev = df.groupby('industry')['annual_revenue_million'].mean().sort_values()

plt.figure(figsize=(10,5))

industry_rev.plot(kind='barh')

plt.title("Average Revenue by Industry")
plt.xlabel("Average Revenue (Million)")
plt.ylabel("Industry")

plt.show()
```

Purpose:
To compare average revenue across industries.

Observation:
Technology companies generated the highest average revenue, while Healthcare and Manufacturing reported comparatively lower average revenue.

Business Insight:
Industry type plays a significant role in revenue generation.

Industry vs Profit Margin
Purpose:
To compare profitability across industries.

Observation:
Some industries generate higher profit margins despite lower revenue levels.

Business Insight:
High revenue does not always imply high profitability.

```python
# Industry VS Profit Margin

industry_profit = df.groupby('industry')['profit_margin_percent'].mean().sort_values()

plt.figure(figsize=(10,5))

industry_profit.plot(kind='barh')

plt.title("Average Profit Margin by Industry")

plt.show()

plt.show()
```

```python
# Country vs Market Rating

country_rating = df.groupby('country')['market_rating'].mean().sort_values()

plt.figure(figsize=(8,5))

plt.hlines(
    y=country_rating.index,
    xmin=0,
    xmax=country_rating.values
)

plt.plot(
    country_rating.values,
    country_rating.index,
    "o"
)

plt.title("Average Market Rating by Country")

plt.show()

```

```python
#Region vs Revenue

region_rev = df.groupby('region')['annual_revenue_million'].mean()

plt.figure(figsize=(7,7))

plt.pie(
    region_rev,
    labels=region_rev.index,
    autopct='%1.1f%%'
)

centre_circle = plt.Circle((0,0),0.70,fc='white')

fig = plt.gcf()
fig.gca().add_artist(centre_circle)

plt.title("Revenue Contribution by Region")

plt.show()
```

```python
#Employee Count vs Revenue

employee_bins = pd.qcut(
    df['employee_count'],
    q=10
)

emp_rev = df.groupby(employee_bins)['annual_revenue_million'].mean()

plt.figure(figsize=(10,5))

plt.plot(
    range(len(emp_rev)),
    emp_rev.values,
    marker='o'
)

plt.title("Employee Count vs Average Revenue")

plt.xlabel("Employee Groups")
plt.ylabel("Average Revenue")

plt.show()
```

What is Multivariate Analysis?

Multivariate Analysis examines the relationship among three or more variables simultaneously. Unlike bivariate analysis, which studies only two variables, multivariate analysis helps identify complex business patterns involving revenue, employee strength, profitability, customer base, industry type, and geographical regions.

The objective of multivariate analysis is to identify:

Revenue drivers
Industry performance
Regional profitability
Customer behavior patterns
Business growth trends


Documentation :  MULTIVARIATE ANALYSIS - 1
Objective : Industry × Revenue × Employee Count

To compare average revenue and employee count across industries.

Observation
Technology companies showed the highest average revenue.
Manufacturing companies maintained larger workforces.
Finance companies generated strong revenue with relatively fewer employees.


Business Insight

Revenue generation is influenced by workforce size, but industry type also plays a crucial role.

```python
# MULTIVARIATE ANALYSIS - 1
# Industry × Revenue × Employee Count

# Which industries have the highest revenue and workforce simultaneously?

industry_analysis = df.groupby('industry').agg({
    'annual_revenue_million':'mean',
    'employee_count':'mean'
}).round(2)

industry_analysis.plot(
    kind='bar',
    figsize=(10,6)
)

plt.title("Industry Revenue and Employee Count Comparison")
plt.ylabel("Average Value")
plt.xticks(rotation=45)

plt.show()
```

Documentation  MULTIVARIATE ANALYSIS - 2
Objective  Region × Market Rating × Profit Margin

To analyze how profitability and market perception vary across regions.

Observation
North America showed strong market ratings.
Profit margins remained relatively stable across regions.
Asia demonstrated competitive profitability.
Business Insight

Market perception and profitability are positively associated in high-performing regions.

```python
 #MULTIVARIATE ANALYSIS - 2
# Region × Market Rating × Profit Margin

region_analysis = df.groupby('region').agg({
    'market_rating':'mean',
    'profit_margin_percent':'mean'
}).round(2)

region_analysis.plot(
    kind='bar',
    figsize=(10,6)
)

plt.title("Region-wise Profitability and Market Rating")
plt.ylabel("Average Value")

plt.show()

```

MULTIVARIATE ANALYSIS - 3
Country × Customer Count × Revenue

Documentation
Objective

To study customer acquisition and revenue generation across countries.

Observation
Countries with larger customer bases generally generated higher revenue.
Revenue growth is strongly linked to customer expansion.

Business Insight

Customer acquisition is a major contributor to revenue growth.

```python
#MULTIVARIATE ANALYSIS - 3
#Country × Customer Count × Revenue

country_analysis = df.groupby('country').agg({
    'customer_count':'mean',
    'annual_revenue_million':'mean'
}).round(2)

country_analysis.plot(
    kind='bar',
    figsize=(10,6)
)

plt.title("Country-wise Customer Base and Revenue")
plt.ylabel("Average Value")

plt.show()
```

MULTIVARIATE ANALYSIS - 4
Industry × Profit Margin × Revenue

Objective

To compare profitability and revenue generation across industries.

Observation
Some industries generated high revenue but moderate profit margins.
Other industries generated lower revenue but maintained strong profitability.

Business Insight

Revenue alone is not sufficient for measuring business performance; profitability must also be considered.

```python
#MULTIVARIATE ANALYSIS - 4
#Industry × Profit Margin × Revenue

profit_revenue = df.groupby('industry').agg({
    'annual_revenue_million':'mean',
    'profit_margin_percent':'mean'
}).round(2)

profit_revenue.plot(
    kind='bar',
    figsize=(10,6)
)

plt.title("Industry Revenue and Profit Margin Comparison")

plt.show()
```

MULTIVARIATE ANALYSIS - 5
Pivot Table Analysis

Objective

To compare average revenue generated by industries across different regions.

Observation

The pivot table clearly highlights regional performance differences within each industry.

Business Insight

Regional market conditions significantly impact revenue generation.

```python
#MULTIVARIATE ANALYSIS - 5
#Pivot Table Analysis

pivot_table = pd.pivot_table(
    df,
    values='annual_revenue_million',
    index='industry',
    columns='region',
    aggfunc='mean'
)

pivot_table
```

```python
pivot_table.plot(
    kind='bar',
    stacked=True,
    figsize=(10,6)
)

plt.title("Industry Revenue Contribution Across Regions")

plt.ylabel("Average Revenue")

plt.show()
```

Multivariate analysis was performed to examine the interaction among multiple business variables simultaneously. The analysis focused on understanding how industry type, employee strength, revenue generation, customer base, market ratings, and profitability influence organizational performance. Grouped bar charts and pivot table analysis were used to identify business patterns across industries, countries, and regions.

The results indicated that Technology companies generally achieved higher revenue levels, while Manufacturing companies maintained larger workforces. Revenue generation showed a strong association with customer acquisition and employee count. Regional analysis revealed variations in profitability and market ratings, highlighting the influence of geographical factors on business performance. Overall, multivariate analysis provided deeper insights into operational efficiency, market competitiveness, and growth opportunities across industries.
